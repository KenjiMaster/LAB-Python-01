### Punto 3

Muestra el nombre del primer producto.
Agrega un nuevo producto al inventario.
Disminuye el stock de “Leche” en 2 unidades (venta).
Calcula el valor total del inventario (precio * stock por producto).
Muestra los productos cuyo stock sea menor a 10.
Aumenta el precio de todos los productos en un 10%.
Redondea los precios con round(p["precio"], 2) para evitar decimales largos.
Busca un producto por nombre (ingresado por el usuario) y muestra el producto encontrado, ten en cuenta que no puede estar.
Elimina un producto del inventario si su stock es 0.
Ordena la lista por precio ascendente.

```python
# Lista de productos en una tienda
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10},
]

print("Primer producto:",inventario[0])
inventario.append({"nombre": "quipitos", "precio": 1200, "stock": 300})
inventario[1]["stock"] -= 2

# Valor de inventario
cuenta =0
for producto in inventario:
  cuenta += producto["precio"]*producto["stock"]
print("Valor de inventario:",cuenta)

print("Productos stock menor a 10")
for producto in inventario:
  if producto["stock"] <= 10:
    print(producto)

#Aumentar precios
for producto in inventario:
  producto["precio"] += producto["precio"]*0.1

#Redondear precios
for producto in inventario:
  producto["precio"] = round(producto["precio"],2)

#Producto por nombre
nombre = input("Ingresa el nombre del producto: ")
encontrado = False
for producto in inventario:
  if producto["nombre"].lower() == nombre.lower():
    print("Producto encontrado:",producto)
    encontrado = True
    break
if not encontrado:
  print("Producto no encontrado")

#Eliminar productos con stock 0
nuevo = []
for producto in inventario:
  if producto["stock"] != 0:
    nuevo.append(producto)
inventario = nuevo
print(inventario)

#Odenar por precio
for i in range(len(inventario)):
  for j in range(i+1,len(inventario)):
    if inventario[i]["precio"] > inventario[j]["precio"]:
      inventario[i],inventario[j] = inventario[j],inventario[i]
print("Inventario ordenado:",inventario)


```

```python
# Tupla: coordenadas (x, y)
punto = (3, 7)
print("Punto:", punto)

#Rectangulo (ancho,alto)
rectangulo = (4,9)
print(f"Area rectangulo: {rectangulo[0]*rectangulo[1]}")
print(f"Perimetro: {rectangulo[0]*2 + rectangulo[1]*2}")

#Intercambio: da error porque las tuplas son inmutables
#rectangulo[0],rectangulo[1] = rectangulo[1],rectangulo[0]

#Calcular distancia recorrida
from math import sqrt
posiciones = [(0, 0), (1, 2), (3, 4)]
distancia = 0
for i in range(1,len(posiciones)):
  x1,y1 = posiciones[i]
  x2,y2 = posiciones[i-1]
  distancia += sqrt((x1-x2)**2 + (y1-y2)**2)
print(f"Distancia total: {distancia:.2f}")

```

```python
def consultar(cuenta):
  print(cuenta)

def depositar(cuenta,monto):
  cuenta["saldo"] += monto
  global movimientos
  movimientos.append({"tipo":"depositar", "monto": monto, "saldo" : cuenta["saldo"]})

def retirar(cuenta,monto):
  if monto > cuenta["saldo"]:
    print("Saldo insuficiente")
  else:
    cuenta["saldo"] -= monto
  global movimientos
  movimientos.append({"tipo":"retirar", "monto": monto, "saldo" : cuenta["saldo"]})

cuenta = {"titular": "Ana", "saldo": 200_000, "movimientos": []}
movimientos = []
menu = f"-------------ATM-----------\n1. Consultar cuenta.\n2. Depositar.\n3. Retirar.\n4. Salir"
print(menu)
operacion = int(input("Elija una operacion: "))
while True:
  if operacion == 1:
    consultar(cuenta)
  elif operacion == 2:
    depositar(cuenta,int(input("Ingresa el monto: ")))
  elif operacion == 3:
    retirar(cuenta,int(input("Ingresa el monto: ")))
  elif operacion == 4:
    print("\n-----Resumen de operaciones----")
    print("|Tipo------|Monto-----|SaldoFinal|")
    for op in movimientos:
      print(f"|{op["tipo"]+"-"*(10-len(op["tipo"]))}|{str(op["saldo"])+"-"*(10-len(str(op["saldo"])))}|{str(op["saldo"])+"-"*(10-len(str(op["saldo"])))}|")
    print("Saliendo..")
    break
  else:
    print("Opcion invalida")
  print(menu)
  operacion = int(input("Elija una operacion: "))
```

### Punto 1

El programa genera un número aleatorio entre 1 y 100 (import random).
El usuario tiene máximo 7 intentos; en cada intento el programa responde "mayor" o "menor".
Lleva el historial de intentos en una lista y muéstralo al final.
Al terminar, pregunta si desea jugar otra vez y lleva un marcador de partidas ganadas y perdidas.
¿Por qué 7 intentos son suficientes si se juega de forma óptima? Relaciónalo con la búsqueda binaria del Ejercicio 5.

```python
from random import randint
numero_random = randint(1,100)
historial = []
intento = 1
partida = 1
victorias =0
derrotas =0

print("Bienvenido al juego de adivinar el numero, tienes 7 intentos")
while True:
  if intento > 7:
    print("Perdiste")
    derrotas += 1
    jugar = input("Deseas volver a jugar (s/n): ")
    if jugar == "s":
      intento = 1
      partida += 1
      continue
    else:
      break
  print(f"Intento: {intento}")
  numero = int(input("Ingresa el numero: "))
  historial.append((partida,intento,numero))
  if numero == numero_random:
    print("Ganaste")
    victorias += 1
    jugar = input("Deseas volver a jugar (s/n): ")
    if jugar == "s":
      intento = 1
      partida += 1
      continue
    else:
      break
  elif numero > numero_random:
    print("El numero es menor")
  else:
    print("El numero es mayor")
  intento += 1
print(f"Partidas ganadas: {victorias}, partidas perdidas: {derrotas}")
print("\nEste es el historial de intentos")
for p,i,n in historial:
  print(f"Partida {p}: Intento - {i}, numero intentado: {n}")

```

```python
def vecinos_vivos(tablero,x,y):
  contador = 0
  dx = [-1,0,1,0,1,-1,1,-1]
  dy = [0,1,0,-1,1,-1,-1,1]
  for i in range(4):
    nx = x + dx[i]
    ny = y + dy[i]
    if 0 <= nx < 10 and 0 <= ny < 10 and tablero[nx][ny]:
      contador += 1
  return contador

def siguiente_generacion(tablero):
  nuevo = tablero.copy()
  for i in range(10):
    for j in range(10):
      v = vecinos_vivos(tablero,i,j)
      if tablero[i][j] == 1 and not (v==2 or v==3):
        nuevo[i][j] = 0
      elif tablero[i][j] == 0 and v ==3:
        nuevo[i][j] = 1
  return nuevo

tablero = []
fila = []
for i in range(10):
  fila.append(0)
for i in range(10):
  tablero.append(fila)

for i in tablero:
  print(fila)

tablero[0][0]=1
tablero[0][1]=1
tablero[1][0]=1
tablero = siguiente_generacion(tablero)
for i in tablero:
  print(fila)
```
